```python
# @title Default title text
# ---------------------------
# 【0】Dependency installation（Colab）
# ---------------------------
!pip -q install kagglehub nltk vaderSentiment yake

# ---------------------------
# 【1】Import libraries
# ---------------------------
import os, re, zipfile, json, random, math, itertools, gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import Sequential, layers
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense, Dropout, Layer
from tensorflow.keras.callbacks import EarlyStopping

from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from sklearn.metrics import confusion_matrix, roc_curve, precision_recall_curve, auc
import yake

# ---------------------------
# 【2】Global random seed & plotting settings
# ---------------------------
SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)
plt.rcParams["figure.figsize"] = (10, 6)

# ---------------------------
# 【3】Download data
# ---------------------------
import kagglehub
path = kagglehub.dataset_download("beaglelee/amazon-reviews-us-beauty-v1-00-tsv-zip")
print("Path to dataset files:", path)

# Unzip & locate TSV file
tsv_file, zip_file = None, None
for fname in os.listdir(path):
    if fname.lower().endswith(".tsv"):
        tsv_file = os.path.join(path, fname)
    if fname.lower().endswith(".zip"):
        zip_file = os.path.join(path, fname)

if tsv_file is None and zip_file is not None:
    with zipfile.ZipFile(zip_file, 'r') as zf:
        members = [m for m in zf.namelist() if m.lower().endswith(".tsv")]
        assert len(members) >= 1, "No TSV file found inside ZIP."
        zf.extract(members[0], path)
        tsv_file = os.path.join(path, members[0])

assert os.path.exists(tsv_file), "TSV file not found."
print("TSV:", tsv_file)

# ---------------------------
# 【4】Read and basic cleaning
# ---------------------------
usecols = [
    "star_rating", "review_body", "review_date",
    "product_title", "verified_purchase", "total_votes", "helpful_votes"
]
df = pd.read_csv(tsv_file, sep='\t', usecols=usecols, quoting=3, on_bad_lines='skip', dtype=str)

# Type conversion
df["star_rating"] = pd.to_numeric(df["star_rating"], errors="coerce")
df["total_votes"] = pd.to_numeric(df["total_votes"], errors="coerce")
df["helpful_votes"] = pd.to_numeric(df["helpful_votes"], errors="coerce")
df["review_body"] = df["review_body"].fillna("").astype(str)

# Keep only rows with star ratings and text
df = df.dropna(subset=["star_rating"])
df = df[df["review_body"].str.strip().str.len() > 0].reset_index(drop=True)
print("Total rows after basic cleaning:", len(df))
display(df.head(3))
# === A. Distribution of Star Ratings ===
if "star_rating" in df.columns:
    plt.figure()
    df["star_rating"].plot(kind="hist",
                           bins=[0.5,1.5,2.5,3.5,4.5,5.5],
                           rwidth=0.9)
    plt.xticks([1,2,3,4,5])
    plt.xlabel("Star Rating"); plt.ylabel("Count")
    plt.title("Distribution of Star Ratings")
    plt.tight_layout(); plt.show()
else:
    print("Warning: column 'star_rating' not found in df.")

# ---------------------------
# 【5】Configuration section
# ---------------------------
# Sample size
MAX_SAMPLES = 120000

# Vocabulary and sequence length
MAX_VOCAB = 40000
MAX_LEN   = 160

# BiLSTM architecture
EMB_DIM    = 128
LSTM_UNITS = 64
DROPOUT    = 0.30
EPOCHS     = 6
BATCH_SIZE = 256

# —— Persistence related (save/load) ——
SAVE_DIR       = "/content/model_ckpt"
MODEL_PATH     = os.path.join(SAVE_DIR, "absa_bilstm_simpleatt.keras")
TOKENIZER_PATH = os.path.join(SAVE_DIR, "tokenizer.json")
CONFIG_PATH    = os.path.join(SAVE_DIR, "inference_config.json")

# Switch: If True and files exist, “load and skip training”; if False (or commented) then train normally and save at the end
LOAD_SAVED_MODEL = False


# Maximum candidate sentences for ABSA (larger is more comprehensive but more time-consuming)
MAX_CANDIDATE_SENT = 120000

# Fusion weight (BiLSTM dominant)
ALPHA = 0.7
# Sentiment thresholds (fusion score ≤ -0.2 = negative, ≥ 0.2 = positive, others neutral)
NEG_TH = -0.2
POS_TH =  0.2

# ---------------------------
# 【6】Define cleaning and serialization
# ---------------------------
CLEAN_RE = re.compile(r"http\S+|www\.\S+|<.*?>|[^a-zA-Z']+")

def clean_text(s: str) -> str:
    """Basic cleaning: lowercase, remove URL/HTML/non-letters, merge spaces"""
    s = s.lower()
    s = CLEAN_RE.sub(" ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

# Labels: Review-level negative/other (for training binary classifier)
df["label_negative"] = (df["star_rating"] <= 2).astype(int)

# —— Stratified sampling (for training review-level binary classifier) ——
# Idea: Sample half from negative (1) and half from non-negative (0), then concatenate.
neg_df = df[df["label_negative"] == 1]
pos_df = df[df["label_negative"] == 0]

if len(df) > MAX_SAMPLES and len(neg_df) > 0 and len(pos_df) > 0:
    n_each = min(MAX_SAMPLES // 2, len(neg_df), len(pos_df))
    df_small = pd.concat([
        neg_df.sample(n=n_each, random_state=SEED),
        pos_df.sample(n=n_each, random_state=SEED)
    ], ignore_index=True)
else:
    # If dataset is small or one class is empty, do not sample (keep safe)
    df_small = df.copy()

print("Sample size:", len(df_small))
print(df_small["label_negative"].value_counts())

# —— ABSA base data only uses 1–2 star negative reviews——
df_absa_base = df[df["star_rating"] <= 2].copy().reset_index(drop=True)
print("ABSA base (1–2 star) rows:", len(df_absa_base))


# Texts and labels
texts  = df_small["review_body"].astype(str).apply(clean_text).tolist()
labels = df_small["label_negative"].astype(int).values

# Training/validation sets
X_train_text, X_val_text, y_train, y_val = train_test_split(
    texts, labels, test_size=0.2, random_state=SEED, stratify=labels
)

# Serialization ensures word→index mapping matches training. If load the model but use a newly fit tokenizer, indices will mismatch and predictions will be distorted.
from tensorflow.keras.preprocessing.text import tokenizer_from_json

if LOAD_SAVED_MODEL and os.path.exists(TOKENIZER_PATH):
    with open(TOKENIZER_PATH, "r") as f:
        tokenizer = tokenizer_from_json(f.read())
    print("Loaded tokenizer from", TOKENIZER_PATH)
else:
    tokenizer = Tokenizer(num_words=MAX_VOCAB, oov_token="<OOV>")
    tokenizer.fit_on_texts(X_train_text)


def to_seq(text_list):
    seq = tokenizer.texts_to_sequences(text_list)
    return pad_sequences(seq, maxlen=MAX_LEN, padding='post', truncating='post')

X_tr = to_seq(X_train_text)
X_va = to_seq(X_val_text)

print("X_tr shape:", X_tr.shape, "X_va shape:", X_va.shape)

# ---------------------------
# 【7】Build BiLSTM + Attention
# ---------------------------
class SimpleAttention(Layer):
    """ Interpretable Attention, used to extract weighted context from BiLSTM hidden states"""
    def build(self, input_shape):
        self.W = self.add_weight(name="att_weight",
                                 shape=(input_shape[-1], 1),
                                 initializer="glorot_uniform",
                                 trainable=True)
        self.b = self.add_weight(name="att_bias",
                                 shape=(input_shape[1], 1),
                                 initializer="zeros",
                                 trainable=True)
        super().build(input_shape)

    def call(self, x):
        # x: [batch, timesteps, features]
        e = tf.keras.backend.squeeze(tf.keras.backend.tanh(tf.keras.backend.dot(x, self.W) + self.b), axis=-1)  # [B, T]
        a = tf.keras.backend.softmax(e)     # Attention weights [B, T]
        a = tf.keras.backend.expand_dims(a) # [B, T, 1]
        context = x * a                      # Weighted
        return tf.keras.backend.sum(context, axis=1)  # [B, F]

# —— Try loading saved model first; if none, then build and train ——
if LOAD_SAVED_MODEL and os.path.exists(MODEL_PATH):
    print(f"Loading saved model from {MODEL_PATH}")
    model = tf.keras.models.load_model(
        MODEL_PATH,
        custom_objects={"SimpleAttention": SimpleAttention}  # Custom layer deserialization
    )
else:
    # === Build + compile ===
    inputs = layers.Input(shape=(MAX_LEN,))
    x = Embedding(input_dim=MAX_VOCAB, output_dim=EMB_DIM)(inputs)
    x = Bidirectional(LSTM(LSTM_UNITS, return_sequences=True))(x)
    x = SimpleAttention()(x)
    x = Dropout(DROPOUT)(x)
    x = Dense(64, activation="relu")(x)
    outputs = Dense(1, activation="sigmoid")(x)

    model = tf.keras.Model(inputs, outputs)
    model.compile(
        loss="binary_crossentropy", optimizer="adam",
        metrics=[tf.keras.metrics.AUC(name="AUC"),
                 tf.keras.metrics.Precision(name="Precision"),
                 tf.keras.metrics.Recall(name="Recall")]
    )
    model.summary()

    early = EarlyStopping(monitor="val_AUC", patience=3, mode="max", restore_best_weights=True)

    history = model.fit(
        X_tr, y_train,
        validation_data=(X_va, y_val),
        epochs=EPOCHS, batch_size=BATCH_SIZE,
        callbacks=[early], verbose=1
    )

    # === After training: save model + tokenizer + key hyperparameters at once ===
    os.makedirs(SAVE_DIR, exist_ok=True)
    model.save(MODEL_PATH)
    with open(TOKENIZER_PATH, "w") as f:
        f.write(tokenizer.to_json())
    with open(CONFIG_PATH, "w") as f:
        json.dump({
            "MAX_LEN": MAX_LEN,
            "MAX_VOCAB": MAX_VOCAB,
            "EMB_DIM": EMB_DIM,
            "ALPHA": ALPHA, "NEG_TH": NEG_TH, "POS_TH": POS_TH
        }, f)
    print("Saved model to:", MODEL_PATH)
    print("Saved tokenizer to:", TOKENIZER_PATH)
    print("Saved config to:", CONFIG_PATH)


# === B. Training Curves ===
if 'history' in locals() or 'history' in globals():
    def plot_metric(h, key, title):
        k, vk = key, "val_" + key
        if k in h.history and vk in h.history:
            plt.figure(); plt.plot(h.history[k], label=k); plt.plot(h.history[vk], label=vk)
            plt.xlabel("Epoch"); plt.ylabel(key); plt.title(title)
            plt.legend(); plt.tight_layout(); plt.show()
        else:
            print(f"Skip: '{key}' not found in history.")
    plot_metric(history, "AUC",       "Training Curve - AUC")
    plot_metric(history, "Precision", "Training Curve - Precision")
    plot_metric(history, "Recall",    "Training Curve - Recall")
    plot_metric(history, "loss",      "Training Curve - Loss")
else:
    print("Skip training curves: loaded a pre-trained model.")

# Validation-set report
y_pred = (model.predict(X_va).ravel() >= 0.5).astype(int)
print(classification_report(y_val, y_pred, digits=4))

# === C. ROC / PR / Confusion Matrix ===
# Compute validation-set probability scores (for ROC/PR)
y_score = model.predict(X_va, batch_size=1024).ravel()
y_pred_cm = (y_score >= 0.5).astype(int)

# ROC
fpr, tpr, _ = roc_curve(y_val, y_score)
roc_auc = auc(fpr, tpr)
plt.figure()
plt.plot(fpr, tpr, label=f"ROC AUC = {roc_auc:.3f}")
plt.plot([0,1], [0,1], linestyle="--")
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend(); plt.tight_layout(); plt.show()

# Precision–Recall
prec, rec, _ = precision_recall_curve(y_val, y_score)
pr_auc = auc(rec, prec)
plt.figure()
plt.plot(rec, prec, label=f"PR AUC = {pr_auc:.3f}")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.title("Precision–Recall Curve")
plt.legend(); plt.tight_layout(); plt.show()

# Confusion Matrix（with numeric annotations）
cm = confusion_matrix(y_val, y_pred_cm)
plt.figure()
plt.imshow(cm, cmap="Blues")
plt.title("Confusion Matrix"); plt.colorbar()
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")
plt.xlabel("Predicted"); plt.ylabel("True")
plt.tight_layout(); plt.show()


# ===================== ABSA Unit (no safety switch, always executed) =====================
# Sentence-level aspect extraction → BiLSTM+VADER sentiment fusion → Aggregation/Visualization/Keywords → Export

import re, random, numpy as np, pandas as pd, matplotlib.pyplot as plt
from IPython.display import display, Markdown

# —— Pre-check variables/functions (ensure standalone execution without NameError)——
required_vars = ["model", "tokenizer", "df_small"]
for v in required_vars:
    assert v in globals(), f"Missing variables `{v}`，Please complete the model training part first."

if "clean_text" not in globals():
    CLEAN_RE = re.compile(r"http\S+|www\.\S+|<.*?>|[^a-zA-Z']+")
    def clean_text(s: str) -> str:
        s = s.lower()
        s = CLEAN_RE.sub(" ", s)
        s = re.sub(r"\s+", " ", s).strip()
        return s

if "to_seq" not in globals():
    from tensorflow.keras.preprocessing.sequence import pad_sequences
    MAX_LEN = int(model.inputs[0].shape[1])
    def to_seq(text_list):
        seq = tokenizer.texts_to_sequences(text_list)
        return pad_sequences(seq, maxlen=MAX_LEN, padding='post', truncating='post')

# —— Sentence splitting (NLTK only) ——
from nltk.tokenize import sent_tokenize

def safe_sent_tokenize(text: str):
    text = str(text).strip()
    return sent_tokenize(text)


# —— Aspect lexicon ——
ASPECT_LEXICON = {
    "quality_durability": [
        r"\bquality\b", r"\bdurable\b", r"\bdurability\b", r"\bbroke(n)?\b", r"\bchip(ped)?\b",
        r"\bstopped working\b", r"\bdefect(ive)?\b", r"\bpoorly made\b", r"\bcheap(ly)?\b"
    ],
    "scent_fragrance": [
        r"\bscent\b", r"\bsmell(s|ed|ing)?\b", r"\bfragrance\b", r"\bperfume\b", r"\bstinky\b", r"\bodor\b"
    ],
    "color_shade": [
        r"\bcolor(s)?\b", r"\bshade(s)?\b", r"\bpigment(ed)?\b", r"\bhue\b", r"\bboxy\b", r"\bpatchy\b"
    ],
    "ingredients_irritation": [
        r"\bingredient(s)?\b", r"\ballerg(y|ic)\b", r"\birrit(ation|ate|ated)\b", r"\brace\b",
        r"\bparaben(s)?\b", r"\bformula\b", r"\bchemical(s)?\b", r"\bsulfate(s)?\b"
    ],
    "application_usage": [
        r"\bapply(ing|ied)?\b", r"\bapplication\b", r"\bbrush\b", r"\bpump\b", r"\bnozzle\b",
        r"\bspread(s|ing)?\b", r"\bblend(s|ing|ed)?\b", r"\bsticky\b", r"\bgreasy\b", r"\boily\b"
    ],
    "longevity_wear": [
        r"\blast(s|ed|ing)?\b", r"\blongevity\b", r"\bwear(s|ing)?\b", r"\bfade(s|d|ing)?\b", r"\btransfer(s|red)?\b"
    ],
    "price_value": [
        r"\bprice(y)?\b", r"\bexpensive\b", r"\boverpriced\b", r"\bcheap\b", r"\bworth\b", r"\bvalue\b", r"\bcost\b"
    ],
    "size_quantity": [
        r"\bsize(d)?\b", r"\bquantity\b", r"\bamount\b", r"\bvolume\b", r"\bml\b", r"\boz\b", r"\btoo small\b"
    ],
    "packaging": [
        r"\bpackage(d|s|ing)?\b", r"\bbox\b", r"\bseal(ed)?\b", r"\bleak(s|ed|ing)?\b", r"\bcap\b"
    ],
    "delivery_shipping": [
        r"\bdeliver(y|ed|ing)\b", r"\bship(ped|ping)?\b", r"\blate\b", r"\bdelay(ed|s)?\b", r"\bdamaged\b"
    ],
    "customer_service": [
        r"\bcustomer service\b", r"\bseller\b", r"\breturn(s|ed)?\b", r"\brefund(ed|s)?\b",
        r"\breplacement\b", r"\bsupport\b"
    ],
    "authenticity": [
        r"\bauthentic\b", r"\bfake\b", r"\bcounterfeit\b", r"\bnot (real|genuine)\b", r"\bexpired\b"
    ],
}
ASPECT_PATTERNS = {a: [re.compile(p, re.IGNORECASE) for p in pats] for a, pats in ASPECT_LEXICON.items()}
def extract_aspects_from_sentence(sent_text: str):
    return {asp for asp, pats in ASPECT_PATTERNS.items() if any(p.search(sent_text) for p in pats)}

# —— Generate candidate sentences ——
MAX_CANDIDATE_SENT = 120000
candidates, candidate_clean_texts = [], []
review_texts = df_absa_base["review_body"].astype(str).tolist()
for ridx, raw in enumerate(review_texts):
    for s in safe_sent_tokenize(raw):
        aspects = extract_aspects_from_sentence(s)
        if aspects:
            candidates.append((ridx, s, aspects))
            candidate_clean_texts.append(clean_text(s))
            if len(candidates) >= MAX_CANDIDATE_SENT:
                break
    if len(candidates) >= MAX_CANDIDATE_SENT:
        break
print("Candidate (sentence, aspects):", len(candidates))
if len(candidates) == 0:
    print("No sentence matched ASPECT_LEXICON. Please enrich aspect patterns and rerun.")
else:
    # —— Sentence-level sentiment: BiLSTM + VADER fusion ——
    X_sent = to_seq(candidate_clean_texts)
    prob_neg = model.predict(X_sent, batch_size=1024, verbose=0).ravel()

    from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
    sia = SentimentIntensityAnalyzer()
    vader_scores = np.array([sia.polarity_scores(s)["compound"] for _, s, _ in candidates])

    ALPHA, NEG_TH, POS_TH = 0.7, -0.2, 0.2
    score_bilstm = 1.0 - 2.0 * prob_neg              # [-1,1], the smaller the more negative
    fusion_score = ALPHA*score_bilstm + (1-ALPHA)*vader_scores

    def label_from_score(sc):
        if sc <= NEG_TH: return "negative"
        if sc >= POS_TH: return "positive"
        return "neutral"
    labels_sent = np.array([label_from_score(s) for s in fusion_score])

    # —— Generate ABSA details ——
    prod_titles = df_absa_base["product_title"].astype(str).tolist() if "product_title" in df_absa_base else [""]*len(df_absa_base)
    stars       = df_absa_base["star_rating"].astype(float).tolist()   if "star_rating"  in df_absa_base else [np.nan]*len(df_absa_base)
    helpful_list = (df_absa_base["helpful_votes"].fillna(0).astype(float).astype(int).tolist() if "helpful_votes" in df_absa_base.columns else [0]*len(df_absa_base))
    total_list   = (df_absa_base["total_votes"].fillna(0).astype(float).astype(int).tolist() if "total_votes"   in df_absa_base.columns else [0]*len(df_absa_base))

    absa_rows = []
    for i, (ridx, sent, aspects) in enumerate(candidates):
        for asp in aspects:
            absa_rows.append({
                "review_index": int(ridx),
                "product_title": prod_titles[ridx],
                "star_rating": float(stars[ridx]) if not pd.isna(stars[ridx]) else np.nan,
                "sentence": sent,
                "aspect": asp,
                "sentiment": labels_sent[i],
                "fusion_score": float(fusion_score[i]),
                "prob_negative_bilstm": float(prob_neg[i]),
                "vader_compound": float(vader_scores[i]),
                "helpful_votes": int(helpful_list[ridx]),
                "total_votes":   int(total_list[ridx]),
            })
    absa_df = pd.DataFrame.from_records(absa_rows)
    print("ABSA rows:", len(absa_df))
    display(Markdown("Sample of Aspect_Sentiment Pairs"))
    display(absa_df.head(3).style.set_caption("Each row = a (sentence, aspect) with fused sentiment"))


    # —— Aspect summary & representative negative sentences ——
    summary = (absa_df
               .groupby("aspect", as_index=False)
               .agg(total=("sentiment","size"),
                    negative=("sentiment", lambda s: (s=="negative").sum())))
    summary["neg_rate"] = summary["negative"]/summary["total"]
    summary = summary.sort_values(["negative","neg_rate"], ascending=[False,False]).reset_index(drop=True)

    display(Markdown("Top Negative Aspects (by Count & Rate)"))
    display(summary.head(12).style.set_caption("Columns: total / negative / neg_rate (sorted)"))


    topK = 3
    neg_only = absa_df[absa_df["sentiment"]=="negative"]
    rep_rows = []
    for asp, g in neg_only.groupby("aspect"):
        g2 = g.sort_values(["fusion_score", "helpful_votes"], ascending=[True, False]).head(topK)  # Avoid “extreme noisy sentences”: sort by fusion_score ascending (more negative first), then by helpful_votes descending (more helpful first)
        for _, r in g2.iterrows():
            rep_rows.append({
                "aspect": asp,
                "example_sentence": r["sentence"],
                "fusion_score": r["fusion_score"],
                "product_title": r["product_title"],
                "star_rating": r["star_rating"],
            })
    representative_examples = pd.DataFrame(rep_rows)
    display(Markdown("Representative Negative Sentences per Aspect (Top 3)"))
    display(representative_examples.head(30).style.set_caption("Sorted by more negative fusion_score, then helpful_votes"))


    # —— Visualization (two plots, one each) ——
    topN = 12
    top_summary = summary.head(topN)

    plt.figure(figsize=(10,6))
    plt.barh(top_summary["aspect"][::-1], top_summary["negative"][::-1])
    plt.xlabel("Negative Count"); plt.ylabel("Aspect"); plt.title("Top Negative Aspects (by count)")
    plt.tight_layout(); plt.show()

    plt.figure(figsize=(10,6))
    plt.barh(top_summary["aspect"][::-1], top_summary["neg_rate"][::-1])
    plt.xlabel("Negative Rate"); plt.ylabel("Aspect"); plt.title("Top Aspects by Negative Rate")
    plt.tight_layout(); plt.show()

    # —— Negative keywords (YAKE) ——
    try:
        import yake
    except ImportError:
        import sys, subprocess
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "yake"], check=False)
        import yake

    kw_extractor = yake.KeywordExtractor(lan="en", n=2, top=10)
    aspect_kw_rows = []
    for asp, g in neg_only.groupby("aspect"):
        texts = g["sentence"].astype(str).tolist()
        sample_n = min(3000, len(texts))
        if sample_n < len(texts):
            texts = random.sample(texts, sample_n)
        all_kws = []
        for t in texts:
            for k, s in kw_extractor.extract_keywords(t):
                all_kws.append(k.lower())
        kw_series = pd.Series(all_kws).value_counts().head(12)
        aspect_kw_rows.append({"aspect": asp, "top_keywords": kw_series.index.tolist()})
    aspect_keywords = pd.DataFrame(aspect_kw_rows).sort_values("aspect").reset_index(drop=True)
    display(Markdown("Top Negative Keywords per Aspect (YAKE)"))
    display(aspect_keywords.style.set_caption("Frequent 1-gram and 2-gram keywords sampled from negative sentences per aspect"))


    # —— Export to CSV (overwrite each time) ——
    out_absa      = "/content/absa_sentence_aspect_results.csv"
    out_summary   = "/content/absa_aspect_summary.csv"
    out_examples  = "/content/absa_aspect_negative_examples.csv"
    out_aspkws    = "/content/absa_aspect_negative_keywords.csv"

    absa_df.to_csv(out_absa, index=False)
    summary.to_csv(out_summary, index=False)
    representative_examples.to_csv(out_examples, index=False)
    aspect_keywords.to_csv(out_aspkws, index=False)

    print("Exported:")
    print(out_absa); print(out_summary); print(out_examples); print(out_aspkws)
# ===================== END ABSA =====================


```
